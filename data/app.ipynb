```python
from langchain_community.document_loaders import PyPDFLoader, TextLoader
from langchain_text_splitters.character import RecursiveCharacterTextSplitter
from sklearn.metrics.pairwise import cosine_similarity

import re 
```

```python
loader = PyPDFLoader("/home/wayne-ramsey/Desktop/CLASSWORK/ollama_basics/data/1706.03762v7.pdf")
pages = loader.load()
```

```python
pages
```

```python
def clean_text(text):
    cleaned_text = re.sub(r'\s+', ' ', text)  # Replace multiple whitespace with a single space
    cleaned_text = re.sub(r'[^a-zA-Z0-9\s.@,!?;:()\-\'"]', '', cleaned_text)  # Remove special characters except punctuation
    return cleaned_text.strip()  # Remove leading and trailing whitespace
```

```python
for page in pages:
    page.page_content = clean_text(page.page_content)
```

```python
pages[0].page_content
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=50,)
```

```python
text = pages[0].page_content
```

```python
chunks = splitter.split_text(text)
```

```python
EMBEDDING_MODEL = "nomic-embed-text-v2-moe"
```

```python
import ollama

response = ollama.embed(
    model=EMBEDDING_MODEL,
    input='my first embedding',
)
print(response.embeddings)
```

```python
text_one = "That car is color blue"
text_two = "The color of that car is blue"
```

```python
response_text_one = ollama.embed(
    model= EMBEDDING_MODEL,
    input=text_one,
)

response_text_two = ollama.embed(
    model= EMBEDDING_MODEL,
    input=text_two,
)
```

```python
print(response_text_one.embeddings)
```

```python
print(response_text_two.embeddings)
```

```python
# literal check fro equality 
response_text_one.embeddings == response_text_two.embeddings
```

```python
cosine_similarity(response_text_one.embeddings, response_text_two.embeddings)
```
